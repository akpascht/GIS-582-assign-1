# Generate a sine wave plot with the following code:
## import numpy and matplotlib python libraries

```python
import numpy as mpy
import matplotlib.pyplot as plt
```

## configure the parameters

```python
# --- Configuration Parameters ---
amplitude = 1.0     # Amplitude of the wave (peak value)
frequency = 2.0     # Frequency in Hertz (cycles per second)
duration = 1.0      # Duration of the wave in seconds
sample_rate = 1000  # How many samples per second (sampling rate)
```

## create a time vector

```python
# --- Create a time vector ---
#    Setting endpoint=False ensures the last point isn't a duplicate of the start of the next cycle if looped.
time = mpy.linspace(0, duration, int(sample_rate * duration), endpoint=False)
```

## calculate the sine wave

```python
# --- Calculate the sine wave values ---
#    np.sin expects the input in radians.
sine_wave = amplitude * mpy.sin(2 * mpy.pi * frequency * time)
```

## lastly, plot the result

```python
# --- Plot the Wave ---
plt.figure(figsize=(10, 4))
plt.plot(time, sine_wave, color='green')
plt.title('Sine Wave')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.grid(True)
plt.show()
```
